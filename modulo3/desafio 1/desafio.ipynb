```python
import pandas as pd

# Carga inicial de datos
df_salaries = pd.read_csv("ds_salaries.csv")
df_salaries.describe()
```

### Pregunta 1
Promedio general de los salarios, desviación estándar, quintiles y rango

```python

mean_usd = df_salaries["salary_in_usd"].mean()
std = df_salaries["salary_in_usd"].std()
q1 = df_salaries["salary_in_usd"].quantile(0.2)
q2 = df_salaries["salary_in_usd"].quantile(0.4)
q3 = df_salaries["salary_in_usd"].quantile(0.6)
q4 = df_salaries["salary_in_usd"].quantile(0.8)
q5 = df_salaries["salary_in_usd"].quantile(1)

disp_range = df_salaries["salary_in_usd"].max() - df_salaries["salary_in_usd"].min()
print("promedio en USD: ", mean_usd)
print("desviación estandar: ", std)
print("percentil 20: ", q1)
print("percentil 40: ", q2)
print("percentil 60: ", q3)
print("percentil 80: ", q4)
print("rango dispersion: ", disp_range)
```

```python
# 2.- Una comparación entre los salarios agrupados por 3 diferentes categorías a tu
# elección. ¿Para cuál de ellas las medidas de tendencia central son más
# representativas? ¿Para cuál son menos representativas? Explica

# primera categoría: tipo de empleo. 
# Agruparemos por nivel de experiencia para analizar los salarios

group_by_type = df_salaries.groupby("employment_type")["salary_in_usd"]

mean_type = group_by_type.mean()
median_type = group_by_type.median()
std_type = group_by_type.std()

print("Promedio salario agrupado por tipo de empleo: ", mean_type)
print("Mediana salario agrupado por tipo de empleo: ", median_type)
print("Desviación estandar agrupado por tipo de empleo: ", std_type)
```

```python
# tercera categoría: tamaño de la compañia. 

group_by_company_size = df_salaries.groupby("company_size")["salary_in_usd"]

mean_company_size = group_by_company_size.mean()
median_company_size = group_by_company_size.median()
std_company_size = group_by_company_size.std()
q1_company_size = group_by_company_size.quantile(0.25)
q3_company_size = group_by_company_size.quantile(0.75)

print("Promedio salario agrupado por tamaño de compañia: ", mean_company_size)
print("Mediana salario agrupado por tamaño de compañia: ", median_company_size)
print("Desviación estandar agrupado por tamaño de compañia: ", std_company_size)
```

```python
# 3.- ¿Qué cargos reciben mejores sueldos en las empresas con sede en Estados Unidos?
# Para ello, primero filtramos por trabajos en US
df_salaries_us = df_salaries[df_salaries["company_location"] == "US"]

# Luego agrupamos por cargo
df_salaries_by_job_title = df_salaries_us.groupby("job_title")

# Y finalmente obtenemos los salarios, ordenados por promedio
df_salaries_us_by_job_title = df_salaries_by_job_title["salary_in_usd"].agg(
    mean='mean',
    median='median',
    max='max'
).sort_values(by="mean", ascending=False)

print(df_salaries_us_by_job_title.head(10))
```
