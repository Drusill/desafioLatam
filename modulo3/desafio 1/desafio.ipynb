## Estadística descriptiva y probabilidades (parte 1)

En este documento realizaremos un análisis estadístico descriptivo de un dataset que contiene información salarial,
por ejemplo, el salario, el cargo, y el tamaño de la compañia de una persona.

```python
import pandas as pd

# Carga inicial de datos
df_salaries = pd.read_csv("ds_salaries.csv")
df_salaries.head()
```

### Pregunta 1
El promedio general de los salarios, desviación estándar, quintiles y rango. (hint: escoge la columna adecuada para comparar)

```python
# Obtendremos esta información a partir de la columna de salarios en dólares

salaries = df_salaries["salary_in_usd"]
mean_usd = salaries.mean()
std = salaries.std()
q1 = salaries.quantile(0.2)
q2 = salaries.quantile(0.4)
q3 = salaries.quantile(0.6)
q4 = salaries.quantile(0.8)
q5 = salaries.quantile(1)
disp_range = salaries.max() - salaries.min()

print("Promedio en USD: ", mean_usd)
print("Desviación estandar: ", std)
print("Percentil 0: ", salaries.min())
print("Percentil 20: ", q1)
print("Percentil 40: ", q2)
print("Percentil 60: ", q3)
print("Percentil 80: ", q4)
print("Percentil 100: ", salaries.max())
print("Rango dispersion: ", disp_range)
```

### Pregunta 2.- 
Una comparación entre los salarios agrupados por 3 diferentes categorías a tu elección. ¿Para cuál de ellas las medidas de tendencia central son más representativas? ¿Para cuál son menos representativas? Explica

##### Primera categoría: Tipo de empleo.

```python
group_by_type = df_salaries.groupby("employment_type")["salary_in_usd"].agg(["mean", "median", "std", "count"])
display(group_by_type)
```

#### Segunda categoría: Ratio de trabajo remoto

```python
group_by_ratio = df_salaries.groupby("remote_ratio")["salary_in_usd"].agg(["mean", "median", "std", "count"])
display(group_by_ratio)
```

#### Tercera categoría: Tamaño de compañía

```python
group_by_company_size = df_salaries.groupby("company_size")["salary_in_usd"].agg(["mean", "median", "std", "count"])
display(group_by_company_size)
```

#### Análisis
Para ello, primero definimos qué significa que las medidas de tendencia central sean o no más representativas.
Si tanto el promedio como su mediana tienen valores similares, y la desviación estándar es baja con respecto al promedio, podemos decir que estas médidas son más representativas de los datos. Es decir, los datos se encuentran más cercas unos de otros en promedio

Entonces:

- Para los tipos de empleo, las medidas de tendencia central en las categorias de freelance (FL) y full time (FT) son más representativas, mientras que en las categorías de contrato (CT) y part time (PT) no lo son.
- En el caso de el ratio de trabajo remoto, las medidas de tendencia central son muy representativas en el caso de trabajo 100% remoto, medianamente representativas para modalidad presencial (0%), y poco representativas para trabajos con modalidad híbrida.
- Por último, las medidas de tendencia central respecto al tamaño de la compañía son más representativas tanto en compañías grandes (L) como medianas (M), mientras que son poco representativas para compañías pequeñas (S).

### Pregunta 3
¿Qué cargos reciben mejores sueldos en las empresas con sede en Estados Unidos? 

Para ello, primero filtramos por trabajos cuyas empresas se encuentran en US.
Luego, agrupamos por el título del cargo, y calculamos el promedio, la mediana y el max del salario en dolares de estos grupos.
Para este conjunto de datos, los cargos mejor pagados son:
- Data analytics lead
- Data science tech lead
- Director of data science

```python

df_salaries_us = df_salaries[df_salaries["company_location"] == "US"]

df_salaries_by_job_title = df_salaries_us.groupby("job_title")
df_salaries_us_by_job_title = df_salaries_by_job_title["salary_in_usd"].agg(
    mean='mean',
    median='median',
    max='max',
    count='count'
).sort_values(by="mean", ascending=False)

display(df_salaries_us_by_job_title.head(10))
```
