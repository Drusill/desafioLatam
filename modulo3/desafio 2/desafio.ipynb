## Estadística descriptiva y probabilidades (parte 2)

En este documento realizaremos una revisión sobre los resultados del Mundial de futbol 2014.

```python
# Cargamos los datos
import pandas as pd

df_worldcup = pd.read_csv('worldcup2014.csv')
df_worldcup.head()
```

### Pregunta 1
Considerando el experimento de seleccionar un país participante al azar, se pide
determinar las probabilidades de los siguientes sucesos:

```python
print("Pertenecer a África: ", df_worldcup['continent'].value_counts(normalize=True)['africa'])
print("Pertenecer a Asia: ", df_worldcup['continent'].value_counts(normalize=True)['asia'])
print("Pertenecer a Europa: ", df_worldcup['continent'].value_counts(normalize=True)['europe'])
print("Pertenecer a Norteamérica: ", df_worldcup['continent'].value_counts(normalize=True)['northamerica'])
print("Pertenecer a Sudamérica: ", df_worldcup['continent'].value_counts(normalize=True)['southamerica'])
```

```python
total = df_worldcup.shape[0]  # guardamos el total de filas
print("Clasificar a segunda ronda: ", df_worldcup[df_worldcup['clasificado'] == 1].shape[0] / total)
print("Ganar al menos un partido: ", df_worldcup[df_worldcup['juegos_ganados'] >= 1].shape[0] / total)
```
