## Estadística descriptiva y probabilidades (parte 2)

En este documento revisaremos los resultados del Mundial de futbol 2014 con el fin de practicar los conceptos de probabilidades.

```python
# Cargamos los datos
import pandas as pd

df_worldcup = pd.read_csv('worldcup2014.csv')
df_worldcup.head()
```

### Pregunta 1
Considerando el experimento de seleccionar un país participante al azar, se pide
determinar las probabilidades de los siguientes sucesos:

```python
# Calculamos la probabilidad de cada suceso como la cantidad de eventos favorables contra el total de eventos
# value_counts cuenta las ocurrencias de cada valor del data frame, en este caso de la columna 'continent'
prob_africa = df_worldcup['continent'].value_counts(normalize=True)['africa']
prob_asia = df_worldcup['continent'].value_counts(normalize=True)['asia']
prob_europa = df_worldcup['continent'].value_counts(normalize=True)['europe']
prob_norteamerica = df_worldcup['continent'].value_counts(normalize=True)['northamerica']
prob_sudamerica = df_worldcup['continent'].value_counts(normalize=True)['southamerica']

print("Pertenecer a África: ", prob_africa)
print("Pertenecer a Asia: ", prob_asia)
print("Pertenecer a Europa: ", prob_europa)
print("Pertenecer a Norteamérica: ", prob_norteamerica)
print("Pertenecer a Sudamérica: ", prob_sudamerica)
```

```python
# Aqui, generamos la condición pertinente de cada suceso, filtramos y calculamos la probabilidad
total = df_worldcup.shape[0]  # guardamos el total de filas
prob_clasificado = df_worldcup[df_worldcup['clasificado'] == 1].shape[0] / total
prob_ganar_un_partido = df_worldcup[df_worldcup['juegos_ganados'] >= 1].shape[0] / total

print("Clasificar a segunda ronda: ", prob_clasificado)
print("Ganar al menos un partido: ", prob_ganar_un_partido)
```

```python
# Ahora, uniremos o intersectaremos sucesos
prob_africa_y_2da_ronda = ((df_worldcup['continent'] == 'africa') & (df_worldcup['clasificado'] == 1)).sum() / total
prob_europa_y_2da_ronda = ((df_worldcup['continent'] == 'europe') & (df_worldcup['clasificado'] == 1)).sum() / total
prob_asia_y_2da_ronda = ((df_worldcup['continent'] == 'asia') & (df_worldcup['clasificado'] == 1)).sum() / total
prob_sudamerica_y_2da_ronda = ((df_worldcup['continent'] == 'southamerica') & (df_worldcup['clasificado'] == 1)).sum() / total
prob_norteamerica_y_2da_ronda = ((df_worldcup['continent'] == 'northamerica') & (df_worldcup['clasificado'] == 1)).sum() / total

prob_africa_o_2da_ronda = ((df_worldcup['continent'] == 'africa') | (df_worldcup['clasificado'] == 1)).sum() / total
prob_europa_o_2da_ronda = ((df_worldcup['continent'] == 'europe') | (df_worldcup['clasificado'] == 1)).sum() / total
prob_asia_o_2da_ronda = ((df_worldcup['continent'] == 'asia') | (df_worldcup['clasificado'] == 1)).sum() / total
prob_sudamerica_o_2da_ronda = ((df_worldcup['continent'] == 'southamerica') | (df_worldcup['clasificado'] == 1)).sum() / total
prob_norteamerica_o_2da_ronda = ((df_worldcup['continent'] == 'northamerica') | (df_worldcup['clasificado'] == 1)).sum() / total

prob_africa_y_ganar_un_partido = ((df_worldcup['continent'] == 'africa') & (df_worldcup['juegos_ganados'] >= 1)).sum() / total
prob_europa_y_ganar_un_partido = ((df_worldcup['continent'] == 'europe') & (df_worldcup['juegos_ganados'] >= 1)).sum() / total
prob_asia_y_ganar_un_partido = ((df_worldcup['continent'] == 'asia') & (df_worldcup['juegos_ganados'] >= 1)).sum() / total
prob_sudamerica_y_ganar_un_partido = ((df_worldcup['continent'] == 'southamerica') & (df_worldcup['juegos_ganados'] >= 1)).sum() / total
prob_norteamerica_y_ganar_un_partido = ((df_worldcup['continent'] == 'northamerica') & (df_worldcup['juegos_ganados'] >= 1)).sum() / total

prob_ganar_un_partido_y_2da_ronda = ((df_worldcup['juegos_ganados'] >= 1) & (df_worldcup['clasificado'] == 1)).sum() / total

print("Pertenecer a África y clasificar a segunda ronda: ", prob_africa_y_2da_ronda)
print("Pertenecer a Europa y clasificar a segunda ronda: ", prob_europa_y_2da_ronda)
print("Pertenecer a Asia y clasificar a segunda ronda: ", prob_asia_y_2da_ronda)
print("Pertenecer a Sudamérica y clasificar a segunda ronda: ", prob_sudamerica_y_2da_ronda)
print("Pertenecer a Norteamérica y clasificar a segunda ronda: ", prob_norteamerica_y_2da_ronda)
print("Pertenecer a África o clasificar a segunda ronda: ", prob_africa_o_2da_ronda)
print("Pertenecer a Europa o clasificar a segunda ronda: ", prob_europa_o_2da_ronda)
print("Pertenecer a Asia o clasificar a segunda ronda: ", prob_asia_o_2da_ronda)
print("Pertenecer a Sudamérica o clasificar a segunda ronda: ", prob_sudamerica_o_2da_ronda)
print("Pertenecer a Norteamérica o clasificar a segunda ronda: ", prob_norteamerica_o_2da_ronda)
print("Pertenecer a África y ganar al menos un partido: ", prob_africa_y_ganar_un_partido)
print("Pertenecer a Europa y ganar al menos un partido: ", prob_europa_y_ganar_un_partido)
print("Pertenecer a Asia y ganar al menos un partido: ", prob_asia_y_ganar_un_partido)
print("Pertenecer a Sudamérica y ganar al menos un partido: ", prob_sudamerica_y_ganar_un_partido)
print("Pertenecer a Norteamérica y ganar al menos un partido: ", prob_norteamerica_y_ganar_un_partido)
print("Ganar al menos un partido y clasificar a segunda ronda: ", prob_ganar_un_partido_y_2da_ronda)

```

### Pregunta 2
A partir de las probabilidades anteriores, y sin realizar conteos, calcular las siguientes probabilidades:

```python
# Ahora, calculamos las probabilidades condicionales utilizando el teorema de Bayes, y las 
# probabilidades calculadas anteriormente

prob_europa_dado_2da_ronda = prob_europa_y_2da_ronda / prob_clasificado
prob_sudamerica_dado_2da_ronda = prob_sudamerica_y_2da_ronda / prob_clasificado
prob_africa_dado_ganar_un_partido = prob_africa_y_ganar_un_partido / prob_ganar_un_partido
prob_ganar_un_partido_dado_2da_ronda = prob_ganar_un_partido_y_2da_ronda / prob_clasificado
prob_ganar_un_partido_dado_norteamerica = prob_norteamerica_y_ganar_un_partido / prob_norteamerica
prob_no_clasificado_dado_africa = (prob_africa - prob_africa_y_2da_ronda) / prob_africa
print("Pertenecer a Europa, si se sabe que clasificó: ", prob_europa_dado_2da_ronda)
print("Pertenecer a Sudamérica, si se sabe que clasificó: ", prob_sudamerica_dado_2da_ronda)
print("Pertenecer a África, si se sabe que ganó al menos un partido: ", prob_africa_dado_ganar_un_partido)
print("Haber ganado al menos un partido, si se sabe que clasificó: ", prob_ganar_un_partido_dado_2da_ronda)
print("Haber ganado al menos un partido, si se sabe que pertenece a Norteamérica: ", prob_ganar_un_partido_dado_norteamerica)
print("No haber clasificado, si se sabe que pertenece a África: ", prob_no_clasificado_dado_africa)
```
